### **2. Overview of the Project**

In this project we are given two different datasets both related to music. In the first dataset, the training set, we have characteristics of different songs such as *id*, *title*, *artist*, *year* (year the song was launched in), *views*, *features*, *lyrics* (the lyrics of the song) and *tag* (the main musical genre of the song). In the second dataset, the test set (that has unseen data), we have the same characteristics of the training set, besides the tag that is missing. What is expect from us is to, by using machine-learning models (Natural Language Processing) to analyze the lyrics of a song and its features, and to be able to predict the musical genre of each song.

### **3. Main Goals of the Project**


We have two main goals in our project.

The first one, related to genre-identification, is to create a classification model, that accurately predicts the genre of a song from its lyrics. The training envolved to get this classification model is also part of this first goal.

The second one, related to sentiment-analysis, is to analyze the emotional undertones within each song lyrics to conclude the emotional essence of each genre. This part is a little bit more complex that the previous one since we need to derive sentimental ratings from words (lyrics) and find notable and useful sentiment difference within and accross genres. Unlike the first goal, this part of the project is open-ended meaning we can formulate and answer as many question as we desire to better conclude the emotions behind each word and to correctly assign it to a certain musical genre.
