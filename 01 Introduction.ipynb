### **2. Overview of the Project**

Song lyrics are one of the most direct ways artists express emotions, thoughts and stories. These lyrical themes and emotional tones are not uniform. They vary across musical genres and are shaped by each genre's culture, conventions and audience. Lyrics are unstructured text, so they are a natural fit for **Natural Language Processing (NLP)**.

In this project we analyse a dataset of songs described by their **title, artist, release year, number of views, featured artists and full lyrics**. The lyrics include section tags such as *Intro* and *Chorus*. The training set also contains each song's **main musical genre** (`tag`). The test set has the same attributes without the genre, and we use it to evaluate our model on unseen data.

The project has two complementary parts:
- a **supervised** part, where we build a model to classify songs by genre from their lyrics;
- an **exploratory** part, where we measure the sentiment in the lyrics and study how it varies within and across genres, over time, and in relation to popularity.

### **3. Main Goals of the Project**

1. **Genre Identification.** Build and evaluate a text classification model that predicts a song's genre from its lyrics. This means comparing different text representations, preprocessing strategies and algorithms, and choosing the best one with proper validation and performance metrics.

2. **Sentiment Analysis.** Derive sentiment scores from the lyrics to describe the emotional profile of each genre. We aim to answer questions such as:
   - What are the predominant sentiments in the lyrics of each genre?
   - Has the predominant sentiment of each genre changed over the years?
   - Does a song's sentiment relate to its popularity (views), within a genre or across genres?

3. **Linking both parts.** Find out which linguistic and emotional features set genres apart, and whether sentiment helps explain why some genres are easier or harder to classify than others.
